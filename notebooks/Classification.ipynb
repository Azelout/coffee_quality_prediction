```python
ALT_MAX = 2800

alt_outliers = df["altitude_mean_meters"] > ALT_MAX
print(f"{alt_outliers.sum()} lots above {ALT_MAX} m -> NaN")
display(df.loc[alt_outliers, ["country_of_origin", "region", "altitude_mean_meters"]])

df.loc[alt_outliers, "altitude_mean_meters"] = np.nan
```

### Feature selection

The target comes from `total_cup_points`, and `total_cup_points` is the sum of the 10 sensory scores
(checked below). So:

- We always remove `total_cup_points`: it gives the answer directly (data leakage).
- If we keep the sensory scores, the model only has to add them, so the problem is too easy.
  With `USE_SENSORY_FEATURES = False`, we predict quality from production data only
  (origin, altitude, variety, process, defects...).
- We also remove `lot_id` (an ID), `farm_name` and `producer` (too many different values for ~1300 lots).

```python
TARGET = "quality_class"
SENSORY_COLS = ["aroma", "flavor", "aftertaste", "acidity", "body", "balance",
                "uniformity", "clean_cup", "sweetness", "cupper_points"]
DROP_COLS = ["lot_id", "total_cup_points", "farm_name", "producer"]

sum_diff = (df[SENSORY_COLS].sum(axis=1) - df["total_cup_points"]).abs()
print(f"|sum of sensory scores - total_cup_points| <= 0.03 for {(sum_diff <= 0.03).sum()}/{len(df)} lots (max {sum_diff.max():.2f})")

USE_SENSORY_FEATURES = False

NUM_COLS = ["harvest_year", "grading_year", "altitude_mean_meters", "moisture_percentage",
            "category_one_defects", "category_two_defects"]
CAT_COLS = ["country_of_origin", "region", "variety", "processing_method", "color"]
if USE_SENSORY_FEATURES:
    NUM_COLS += SENSORY_COLS

# every column is either used, deliberately dropped, or the target
assert set(df.columns) == set(NUM_COLS + CAT_COLS + SENSORY_COLS + DROP_COLS + [TARGET])

X = df[NUM_COLS + CAT_COLS]
y = df[TARGET]
```

### Train, validation and test sets

- Test set (20 %): kept apart and used only at the end.
- Validation: 5-fold cross-validation on the training set (80 %). The dataset is small
  (1337 lots, only 21 `Exemplary` in the test set), so a separate validation set would be too small
  to give stable results.
- All splits are stratified: each set keeps the same class proportions.

```python
from sklearn.base import clone
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.metrics import classification_report, ConfusionMatrixDisplay
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from matplotlib.ticker import PercentFormatter
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)

print(pd.DataFrame({
    "train": y_train.value_counts(normalize=True),
    "test": y_test.value_counts(normalize=True),
}).round(3))
```

### Feature analysis (training set only)

```python
train_df = X_train.join(y_train)

CLASS_ORDER = ["Commercial / Sub-Grade (<80)", "Specialty (80-84)", "Exemplary (85+)"]
CLASS_SHORT = ["Commercial", "Specialty", "Exemplary"]
CLASS_COLORS = dict(zip(CLASS_ORDER, ["#2a78d6", "#eb6834", "#1baf7a"]))

counts = y_train.value_counts().reindex(CLASS_ORDER)

fig, ax = plt.subplots(figsize=(8, 2.8))
bars = ax.barh(CLASS_SHORT, counts, color=[CLASS_COLORS[c] for c in CLASS_ORDER], height=0.6)
ax.bar_label(bars, labels=[f"{n} ({n / counts.sum():.0%})" for n in counts], padding=4)
ax.invert_yaxis()
ax.set_xlim(0, counts.max() * 1.18)
ax.set_xlabel("Number of lots")
ax.set_title("Target distribution (training set)", loc="left")
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()
```

Only 8 % of the lots are `Exemplary` and 14 % are `Commercial`. A model can get 78 % accuracy

```python
n_rows = int(np.ceil(len(NUM_COLS) / 3))
fig, axes = plt.subplots(n_rows, 3, figsize=(14, 3.2 * n_rows), sharey=True)

for ax, col in zip(axes.flat, NUM_COLS):
    sns.boxplot(data=train_df, x=col, y=TARGET, hue=TARGET, order=CLASS_ORDER, hue_order=CLASS_ORDER,
                palette=CLASS_COLORS, legend=False, width=0.6, linewidth=1, fliersize=2, ax=ax)
    if "defects" in col:
        ax.set_xscale("symlog")  # very skewed: most lots have 0-5 defects, a few have 50+
        ax.set_xlim(left=0)
    ax.set_title(col, loc="left")
    ax.set_xlabel("")
    ax.set_ylabel("")
    ax.set_yticks(range(len(CLASS_SHORT)), CLASS_SHORT)
    ax.spines[["top", "right"]].set_visible(False)
for ax in axes.flat[len(NUM_COLS):]:
    ax.set_visible(False)

fig.suptitle("Numeric features by class (training set)", x=0.01, ha="left")
plt.tight_layout()
plt.show()
```

- Altitude is the clearest signal: better coffees grow higher (median ≈ 1250 m for `Commercial`,
  1350 m for `Specialty`, 1630 m for `Exemplary`).
- Defects: `Commercial` lots have more defects (median of 4 category-two defects, against 2 for the other classes).
- Moisture and the years look almost the same for all classes: they are probably weak features.

```python
missing_by_class = (
    train_df[NUM_COLS + CAT_COLS].isna()
    .groupby(train_df[TARGET]).mean()
    .loc[CLASS_ORDER].T * 100
).round(1)
missing_by_class.columns = CLASS_SHORT

print("% of missing values per class (training set)")
missing_by_class[missing_by_class.sum(axis=1) > 0]
```

`Exemplary` lots have more missing values (e.g. `color`: 40 % against 19 % for `Specialty`;
`processing_method`: 25 % against 13 %). So a missing value is a useful signal. This is why the pipeline
keeps this information (missing indicators and `"Unknown"` category) instead of deleting these rows.

```python
def class_share(col, top=None):
    """Share of each class for every value of `col` (NaN -> "Unknown", rare values -> "Other")."""
    values = train_df[col].fillna("Unknown")
    if top:
        keep = values.value_counts().head(top).index
        values = values.where(values.isin(keep), "Other")
    share = pd.crosstab(values, train_df[TARGET], normalize="index")[CLASS_ORDER]
    share = share.sort_values(CLASS_ORDER[-1])
    n = values.value_counts()
    share.index = [f"{v} (n={n[v]})" for v in share.index]
    return share

shares = {
    "country_of_origin (top 10)": class_share("country_of_origin", top=10),
    "processing_method": class_share("processing_method"),
    "color": class_share("color"),
}

fig, axes = plt.subplots(len(shares), 1, figsize=(10, 11),
                         gridspec_kw={"height_ratios": [len(s) for s in shares.values()]})
for ax, (title, share) in zip(axes, shares.items()):
    share.plot.barh(stacked=True, ax=ax, width=0.75, legend=False,
                    color=[CLASS_COLORS[c] for c in CLASS_ORDER], edgecolor="white", linewidth=1.5)
    ax.set_title(title, loc="left")
    ax.set_xlim(0, 1)
    ax.xaxis.set_major_formatter(PercentFormatter(1))
    ax.spines[["top", "right"]].set_visible(False)

handles, _ = axes[0].get_legend_handles_labels()
fig.legend(handles, CLASS_SHORT, loc="upper right", ncol=3, frameon=False)
fig.suptitle("Share of each class by category (training set)", x=0.01, ha="left")
plt.tight_layout(rect=(0, 0, 1, 0.97))
plt.show()
```

- Country is very informative: 54 % of the lots from Ethiopia are `Exemplary`,
  while 30 % of the lots from Mexico are `Commercial`.
- `processing_method` and `color` have a weaker effect. The `Unknown` category has more `Exemplary` lots,
  as seen in the table above.

### Preprocessing pipeline

- Numeric columns: missing values are replaced by the median (less sensitive to outliers than the mean).
  We also add a column that says if the value was missing, because missing data can carry information.
  Then we scale the values.
- Text columns: missing values become an `"Unknown"` category, then one-hot encoding.
  Categories with fewer than 10 lots are grouped together (useful for `region`, which has 356 values).

The pipeline is fitted on the training set only, so the test set does not influence
the preprocessing (no data leakage).

```python
numeric_pipe = Pipeline([
    ("impute", SimpleImputer(strategy="median", add_indicator=True)),
    ("scale", StandardScaler()),
])
categorical_pipe = Pipeline([
    ("impute", SimpleImputer(strategy="constant", fill_value="Unknown")),
    ("onehot", OneHotEncoder(handle_unknown="infrequent_if_exist", min_frequency=10, sparse_output=False)),
])
preprocess = ColumnTransformer([
    ("num", numeric_pipe, NUM_COLS),
    ("cat", categorical_pipe, CAT_COLS),
]).set_output(transform="pandas")

X_train_prep = preprocess.fit_transform(X_train)
print(X_train_prep.shape)
X_train_prep.head()
```

## Baseline models

We use two simple baselines:

1. Dummy classifier (`most_frequent`): always predicts `Specialty`. It learns nothing and gives
   the minimum score to beat.
2. Logistic regression with `class_weight="balanced"`: a simple and fast linear model.
   The class weights give more importance to the rare classes, so the model does not ignore them.
   It is the real reference for the next, more complex models.

Metrics: the classes are imbalanced, so accuracy is misleading. Our main metric is the
macro F1-score, which gives the same weight to each class. We also report the balanced accuracy
(mean recall of the classes), and precision / recall for each class.

```python
models = {
    "Dummy (most frequent)": DummyClassifier(strategy="most_frequent"),
    "Logistic regression (balanced)": LogisticRegression(class_weight="balanced", max_iter=2000),
}
pipelines = {
    name: Pipeline([("preprocess", clone(preprocess)), ("clf", clf)])
    for name, clf in models.items()
}
```

### Cross-validation (training set)

5-fold stratified cross-validation. The whole pipeline (preprocessing included) is fitted again on each fold,
so there is no leakage from the validation fold. We use these scores to compare models.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
SCORING = {"F1 macro": "f1_macro", "Balanced accuracy": "balanced_accuracy", "Accuracy": "accuracy"}

cv_results = {}
for name, pipe in pipelines.items():
    scores = cross_validate(pipe, X_train, y_train, cv=cv, scoring=SCORING)
    cv_results[name] = {
        metric: f"{scores[f'test_{metric}'].mean():.3f} ± {scores[f'test_{metric}'].std():.3f}"
        for metric in SCORING
    }

pd.DataFrame(cv_results).T
```

### Test set evaluation

Each model is trained on the whole training set and evaluated once on the test set.

```python
fig, axes = plt.subplots(1, len(pipelines), figsize=(6 * len(pipelines), 4.5))

for ax, (name, pipe) in zip(axes, pipelines.items()):
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test)

    print(f"=== {name}")
    print(classification_report(y_test, y_pred, labels=CLASS_ORDER, target_names=CLASS_SHORT, zero_division=0))

    ConfusionMatrixDisplay.from_predictions(
        y_test, y_pred, labels=CLASS_ORDER, display_labels=CLASS_SHORT,
        cmap="Blues", colorbar=False, ax=ax,
    )
    ax.set_title(name, loc="left")

fig.suptitle("Confusion matrices (test set)", x=0.01, ha="left")
plt.tight_layout()
plt.show()
```

### Initial results

| Model | F1 macro (CV) | F1 macro (test) | Balanced accuracy (CV) |
|---|---|---|---|
| Dummy | 0.29 | 0.29 | 0.33 |
| Logistic regression | 0.46 ± 0.03 | 0.52 | 0.60 ± 0.05 |

- The dummy model gets 78 % accuracy but never finds `Commercial` or `Exemplary`.
- The logistic regression finds most of the rare lots (recall: 0.78 for `Commercial`, 0.67 for `Exemplary`),
  but it often puts `Specialty` lots in the other classes: 92 of the 210 `Specialty` lots are misclassified.
  Its accuracy is lower (60 %); this is expected with balanced class weights.
- The test score (0.52) is a bit higher than the CV score (0.46). The test set is small (21 `Exemplary` lots),
  so this difference is probably noise.
- Next step: new models must beat a macro F1 of about 0.46 in cross-validation.
  From now on, we compare models with cross-validation and use the test set only for the final model.
